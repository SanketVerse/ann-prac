```python
from numpy import array, random, dot
from random import choice
from matplotlib import pyplot as plt

def step_function(x):
  return 0 if x < 0 else 1
def train_perceptron(training_dataset, learning_rate=0.2,n_iterations=100, fig_title="Error Plot"):
  weights = random.rand(3)
  error = []

  for j in range(n_iterations):
    x, expected = choice(training_dataset)
    result = dot(weights, x)
    err = expected - step_function(result)
    error.append(err)
    weights += learning_rate * err * x

  for x, _ in training_dataset:
    result = dot(x, weights)
    print(f"{x[:2]}: {result} -> {step_function(result)}")

  plt.figure(figsize=(7, 3))
  plt.ylim([-1, 1])
  plt.plot(error)
  plt.title(fig_title)
  plt.xlabel("Iterations")
  plt.ylabel("Error")
  plt.show()
  return weights
```

```python
or_dataset = [
  (array([0,0,1]), 0),
  (array([0,1,1]), 1),
  (array([1,0,1]), 1),
  (array([1,1,1]), 1),
]
train_perceptron(or_dataset, learning_rate=0.2,
fig_title=
"OR Gate Error Plot");
```

```python
and_dataset = [
  (array([0,0,1]), 0),
  (array([0,1,1]), 0),
  (array([1,0,1]), 0),
  (array([1,1,1]), 1),
]
train_perceptron(and_dataset, learning_rate=0.2, fig_title=
"AND Gate Error Plot");
```

```python
learning_rates = [0.2, 0.4, 0.6, 0.8, 1]
print("===== OR Gate =====")
for lr in learning_rates:
  print(f"\nLearning Rate = {lr}")
  train_perceptron(or_dataset, learning_rate=lr, fig_title=f"OR Gate Error Plot with Learning Rate : {lr}");
  print("===== AND Gate =====")
for lr in learning_rates:
  print(f"\nLearning Rate = {lr}")
  train_perceptron(and_dataset, learning_rate=lr,fig_title=f"AND Gate Error Plot with Learning Rate : {lr}");
```
